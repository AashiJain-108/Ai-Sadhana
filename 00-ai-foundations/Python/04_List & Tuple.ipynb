```python
#took example from python.org tutorial 5 to Summarize list topic

fruits = ['orange', 'apple', 'pear', 'banana', 'kiwi', 'apple', 'banana']
fruits.count('apple')

fruits.count('tangerine')

fruits.index('banana')

fruits.index('banana', 4)  # Find next banana starting at position 4

fruits.reverse()
fruits

fruits.append('grape')
fruits

fruits.sort()
fruits

fruits.pop()
```

```python
#Using Lists as Queues
from collections import deque
queue = deque(["Eric", "John", "Michael"])
queue.append("Terry")           # Terry arrives
queue.append("Graham")          # Graham arrives
queue.popleft()                 # The first to arrive now leaves

queue.popleft()                 # The second to arrive now leaves

queue                           # Remaining queue in order of arrival
```

Created method summary
- append → add to end
- extend → add multiple
- insert → specific position
- remove → value
- pop → position + gives removed value back
- index → position of value
- count → frequency
- sort → ordering
- reverse → reverse order
- copy → duplicate list

**Remember:**
- List are mutable 
- Tuple are immutable

## <div align="center"><u>**TUPLE**</u></div>


- A tuple is a collection of objects which ordered and immutable.
-  Tuples are sequences, just like lists. 

- The differences between tuples and lists are, the tuples cannot be changed unlike lists 
- tuples use parentheses(), whereas lists use square brackets[].

 -- Once a tuple is created, we cannot change its values. We cannot use add, insert, remove methods in a tuple because it is not modifiable (mutable). Unlike list, tuple has few methods. Methods related to tuples:

- tuple(): to create an empty tuple
- count(): to count the number of a specified item in a tuple
- index(): to find the index of a specified item in a tuple
- '+' operator: to join two or more tuples and to create a new tuple

```python
empty_tuple = ()
# or using the tuple constructor
empty_tuple = tuple()
print(empty_tuple)
```

```python
tuple1 = ("School","College","Job")
print(tuple1)
```

```python
len(tuple1)
```

**Accessing Tuple Items**

```python
programming_languages = ("Python", "C++", "Java", "Go")

print("Languages:", programming_languages)
print("Type:", type(programming_languages))
```

```python
mixed_tuple = ("Python", 25, 3.14, True)
print("Mixed tuple:", mixed_tuple)
```

```python
#positive indexing

subjects = ("Python", "ML", "DL", "DSA")

print("First subject:", subjects[0])
print("Second subject:", subjects[1])
print("Last subject:", subjects[len(subjects) - 1])
```

```python
#Negative Indexing

print("Last subject:", subjects[-1])
print("Second-last subject:", subjects[-2])
print("First subject:", subjects[-4])
```

```python
topics = ("Python","Statistics","Machine Learning","Deep Learning","NLP")

print("All topics:", topics[0:5])
print("All topics:", topics[:])

print("First three:", topics[:3])
print("Last three:", topics[-3:])

print("Middle topics:", topics[1:4])
print("From second topic:", topics[1:])
```

```python
topics[0] = "Tuple"
```

**Changing Tuples to Lists**

```python
languages = ("Python", "Java", "C++")

language_list = list(languages) 

print("Tuple:", languages)
print("List:", language_list)

print(type(languages))
print(type(language_list))
    
```

```python
updated_languages = tuple(language_list)
print("Updated tuple:", updated_languages)
```

**Tuple Unpacking**

```python
student = ("Aarav", 21, "Computer Science")

name, age, branch = student

print("Name:", name)
print("Age:", age)
print("Branch:", branch)
```

**Extended Unpacking**

```python
marks = (78, 85, 91, 88, 95)

first_mark, *remaining_marks = marks

print("First mark:", first_mark)
print("Remaining marks:", remaining_marks)
```

```python
print(max(marks))
```

```python
print(min(marks))
```
