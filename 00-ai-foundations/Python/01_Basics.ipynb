### Variable Naming Conventions

- **Rule 1:** You should start variable name with an alphabet or underscore (`_`) character.  
- **Rule 2:** A variable name can only contain `A-Z`, `a-z`, `0-9`, and underscore (`_`).  
- **Rule 3:** You cannot start the variable name with a number.  
- **Rule 4:** You cannot use special characters with the variable name such as `$`, `%`, `#`, `&`, `@`, `.`, `-`, `^`, etc.  
- **Rule 5:** Variable names are case sensitive. For example, `str` and `Str` are two different variables.  
- **Rule 6:** Do not use reserved keywords as variable names. For example: `class`, `for`, `def`, `del`, `is`, `else`, `try`, `from`, etc.

**Getting User Input**

 What is User Input? 
- User input refers to data entered by the user during program execution. In Python, you can ask the user to type something, and the program will capture and use that input. 

taking input

```python
name = input("Enter your name:")
print("Welcome!",name)
```

combining inputs

- \+ → concatenates tightly (no space unless you add it).

- , → prints with a space automatically between items.

```python
first_name = input("Enter your first name")
country = input("Enter your country name")

print("Hello",first_name,"from",country)
```

```python
first_name = input("Enter your first name")
country = input("Enter your country name")

print("Hello"+first_name+"from"+country)
```

 Creating a Username

```python
first = input("Enter your first name: ")
year = input("Enter your birth year: ")

username = first.lower() + "_" + year

print("Your username is:", username)
```
