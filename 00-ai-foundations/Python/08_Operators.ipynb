| Category   | Operators                       |
| ---------- | ------------------------------- |
| Arithmetic | `+ - * / // % **`               |
| Comparison | `> < >= <= == !=`               |
| Logical    | `and or not`                    |
| Assignment | `= += -= *= /= //= %= **=` etc. |
| Bitwise    | `& \| ^ ~ << >>`                |
| Membership | `in`, `not in`                  |
| Identity   | `is`, `is not`                  |


````python
# Arithmetric Operators
print("Addition",2+3)

print("subtraction",2-3)

print("multiplication",2*3)

print("division",8/2)

print("floor division",8//2)

print("modulus or remainder",8%2)

print("power",8**2)
````

````python
# Relational Operators
print("greater than",7>5)

print("smaller than",7<5)

print("greater than or equal to",7>=7)

print("smaller than or equal to",7<=7)

print("equal to",7==7)

print("is not equal to",7!=7)
````

````python
# Logical Operators
print("And",1 and 0)

print("Or",1 or 0)

print("Not",not 1)
````

````python
# Bitwise Operators

# bitwise and
print(2 & 3)

# bitwise or
print(2 | 3)

# bitwise xor
print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
````

## Assignment Operators

| Operator | Meaning |
|----------|---------|
| `=` | Assignment |
| `+=` | Addition assignment |
| `-=` | Subtraction assignment |
| `*=` | Multiplication assignment |
| `/=` | Division assignment |
| `//=` | Floor division assignment |
| `%=` | Modulus assignment |
| `**=` | Exponentiation assignment |
| `&=` | Bitwise AND assignment |
| `|=` | Bitwise OR assignment |
| `^=` | Bitwise XOR assignment |
| `<<=` | Left shift assignment |
| `>>=` | Right shift assignment |

Python does not have ++ or -- operators.

Use:

x += 1
x -= 1

````python
# Arithmetic Assignment Operators
x = 10

x += 5
x -= 2
x *= 3
x /= 2
x //= 2
x %= 2
x **= 2

# Bitwise Assignment Operators
y = 10

y &= 3
y |= 3
y ^= 3
y <<= 1
y >>= 1
````

````python
#membership operator(in/not in)
print("Python" in "I am learning Python")
print("Java" not in "I am learning Python")
print('C' not in 'Coder')
print(1 in [2,3,4,5,6])
````

````python
a = [1, 2, 3]
b = a

print(a is b)
````

````python
#Identity operator (liked this example from w3 school)
x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is y)
print(x is not y)
````

- == asks: "Do these two objects contain the same value?"
- is asks: "Are these two variables pointing to the exact same object?"

== → compare values

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
````

is → compare identity

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a is b)
````

**asks**

> Is `a` pointing to the exact same object as `b`?
