```python
# Student will get approved if he will get avg grade more than or equal to 6 and 
# His attendence is more than and equal to 80% 
 
grade1=float(input("Type the grade of first test: ")) 
grade2=float(input("Type the grade of Second test: ")) 
absences=int(input("Type the number of absences: ")) 
total_classes=int(input("Type total no of classes: ")) 
 
avg_grade=(grade1+grade2)/2 
attendance=(total_classes-absences)/total_classes 
 
if(avg_grade>=6): 
    if(attendance>=0.8): 
        print("Student has been approved") 
    else: 
        print("Student has failed due to attendance rate lower than 80% ") 
else: 
    print("Student has failed. ") 
```

```python
# Program: Find the sum of digits of a 3-digit number

try:
    number = int(input("Enter a 3-digit number: "))

    if 100 <= number <= 999:
        a = number % 10
        number = number // 10

        b = number % 10
        number = number // 10

        c = number % 10

        print("Sum of digits:", a + b + c)

    else:
        print("Error: Please enter exactly a 3-digit number.")

except ValueError:
    print("Error: Please enter a valid integer.")
```
