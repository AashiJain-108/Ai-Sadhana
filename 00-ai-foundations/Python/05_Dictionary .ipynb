## <div align="center"><u>**DICTIONARY**</u></div>


- Dictionaries are used to store data values in key: value pairs.
- A dictionary is a collection which is ordered, changeable and does not allow duplicates.
- Keys are unique within a dictionary while values may not be. The values of a dictionary can be of any type, but the keys must be of an immutable data type such as strings, numbers, or tuples. 
- An empty dictionary without any items is written with just two curly braces, like this: { }. 

```python
person = ["Ruhani","Gupta","2005","India"] #this is list
```

```python
#Each element of dictionary is composed by key and value pair
candidate = {"first_name":"Ruhani","last_name":"Gupta","birth_year":"2005","country_of_birth":"India"}
type(candidate)
```

```python
candidate["first_name"]
```

```python
dict1 = {'brand':'Lenovo Thinkpad','model':'x1 carbon','gen':'13','year':'2026'}
print(dict1)
```

```python
x =dict1["model"]
print(x)
```

```python
y = dict1.get("model")
print(y)
```

**Dictionary in python is also mutable, i.e we can change value of a property**

```python
candidate["birth_year"] = 2006
candidate
```

```python
 # We can also add new keys to dictionary 
candidate["nationality"]= "NRI"
candidate
```

**The values in dictionaries are any type, we can have string, number, list, tuple even dictionaries**

```python
candidate["hobbies"] = ["swimming","vollyball"]
candidate
```

```python
candidate["hobbies"].append("hiking")
candidate
```

**Accessing values in Dictionary**

```python
dict = {'Name':'Kangna','Age':13,'Class':'Seventh'}
print("dict['Name']:",dict['Name'])
print("dict['Age']:",dict['Age'])
```

**Update Dictionary**

```python
dict = {'Name':'Kangna','Age':13,'Class':'Seventh'}
dict['Age'] = 14 #update existing entry
dict['School'] = "DPS School" #Add new entry

print("dict['School']:",dict['School'])
print("dict['Age']:",dict['Age'])
```

**Delete Dictionary Elements**

```python
del dict['Name'] #remove entry with key 'Name'
dict.clear() # remove all entries in dict
del dict #delete entire dictionary
print(dict)
```

**Properties of dictionary**

(a) More than one entry per key not allowed. Which means no duplicate key is allowed. When duplicate keys encountered during assignment, the last assignment wins

```python
dict1 = {'Name': 'Zara', 'Age': 7, 'Name': 'Manni'} 
print ("dict1['Name']: ", dict1['Name']) 
```

(b) Keys must be immutable. Which means you can use strings, numbers or tuples as 
dictionary keys but something like ['key'] is not allowed.

```python
dict1 = {['Name']: 'Zara', 'Age': 7} 
print ("dict1['Name']: ", dict1['Name']) 
```
