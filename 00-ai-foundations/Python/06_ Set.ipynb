## <div align="center"><u>**SET**</u></div>


SET → unique collection, unordered

````python
fruits = {'apple', 'banana', 'apple', 'orange'} #in set duplicate apple would be printed once only
print(fruits)
````

**creating a set**

````python
x = set()
````

**size of set**

````python
fruits = {'apple', 'banana', 'orange'}
print(len(fruits))
````

**membership function in set**

````python
print('apple' in fruits)
````

````python
print('mango' in fruits)
````

````python
visited = {'Delhi', 'Mumbai', 'Pune'}

if 'Delhi' in visited:
    print("Already visited")
````

**adding items** : methods
- .add() → one item
- .update() → multiple items

````python
fruits.add('mango')
print(fruits)
````

````python
fruits.update(['grapes', 'kiwi'])
print(fruits)
````

````python
#if an item is already there we try to add it, item gets ignored
fruits.add('kiwi')
print(fruits)
````

**Removing an item**

- remove()   → remove it; complain if absent
- discard()  → remove it; don't complain if absent
- pop()      → It removes and returns an arbitrary element.

````python
fruits.remove('kiwi')
````

````python
fruits.remove('pineapple') #complains about absense of asked item in set
````

````python
fruits.discard('pineapple') #nothing happens no complaint
````

````python
removed = fruits.pop()
print(fruits)
````

- clear() → empty the set
- del     → delete the variable/set entirely

- clear = empty the box 📦
- del = throw away the box 🗑️

````python
fruits.clear()
print(fruits)
````

````python
del fruits
print(fruits)
````

- `clear()` removes all items from the set, but the set variable still exists as an empty set.
- `del` removes the variable itself, so the set no longer exists.

Conversion from list to set 

- list → set
is a simple way to remove duplicates.

````python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = set(numbers) # list to set
print(unique_numbers)
numbers = list(unique_numbers) # back to list
print(numbers)
````

```md
├── Relationships
│   ├── A | B             → union
│   ├── A & B             → intersection
│   ├── A - B             → A only
│   └── A ^ B             → non-common elements
│
├── Containment
│   ├── A.issubset(B)
│   └── A.issuperset(B)
│
└── Relationship
    └── A.isdisjoint(B)   → no common elements?
```


````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
````

**Union**

````python
print(A | B)
print(A.union(B))
````

**Intersection**

````python
print(A & B)
print(A.intersection(B))
````

**Difference**

````python
print(A - B)
print(B - A)
````

**Symmetric difference** : Everything except the things they have in common.

````python
print(A ^ B)
````

**Subset and superset**

````python
C = {1, 2, 3, 4, 5}
D = {2, 3}
````

````python
print(D.issubset(C))
````

````python
C.issuperset(D)
````

**isdisjoint()** : Do these two sets have ZERO elements in common?

````python
A = {1, 2, 3}
B = {4, 5, 6}

A.isdisjoint(B)
````

````python
A = {1, 2, 3}
B = {3, 4, 5}

A.isdisjoint(B)
````
