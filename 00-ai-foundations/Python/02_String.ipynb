## <div align="center"><u>**String**</u></div>


Strings in Python are identified as a contiguous set of characters represented in the quotation marks. Python allows for either pairs of single or double quotes. 

**1.Creating strings**

```python
language = "Python" #double quotes
print(language)

topic = 'String' #single quotes
print(topic)
```

**2. Indexing**

```python
word = "Machine_Learning"

print(word[0])
print(word[-3])
print(word[2])
```

**3. Slicing**

<pre>
[start : stop]
   ↑       ↑
included  excluded
</pre>


```python
word = "MachineLearning"

print(word[:7])
print(word[7:])
print(word[-8:])
print(word[0:7])
```

**4. Length**

```python
course = "Artificial Intelligence"
print(len(course))
```

**5. Strings are immutable**

```python
word = "Python"

# This does NOT work:
 word[0] = "J"
```

```python
word = "J" + word[1:]
print(word)
```

**6. Concatenation**

```python
first_name = "sleepy"
last_name = "coder"

full_name = first_name + " " + last_name

print(full_name)
```

```python
age = 76
print("Age: " + str(age))
```

**7.More methods**

```python
text = "  Python Is Fun  "

print(text.lower())
print(text.upper())
print(text.strip())
print(text.replace("Fun", "Powerful"))
print(text.split())
```

```python
#startswith() checks whether a string begins with specific text.

website = "https://github.com"

print("Starts with https:", website.startswith("https"))
print("Starts with http:", website.startswith("http"))
print("Starts with github:", website.startswith("github"))
```

```python
#endswith() checks whether a string ends with specific text.

filename = "machine_learning.ipynb"

print("Ends with .ipynb:", filename.endswith(".ipynb"))
print("Ends with .py:", filename.endswith(".py"))
print("Ends with notebook:", filename.endswith("notebook"))
```

```python
#join() combines multiple strings using a separator.

languages = ["Python", "C++", "Java"]
combined_languages = " | ".join(languages)
print("Languages:", combined_languages)

#Joining words with a space

words = ["I", "am", "learning", "AI"]
sentence = " ".join(words)
print("Sentence:", sentence)
```

```python
text_1 = "python programming"
text_2 = "hELLO wORLD"

print(text_1.capitalize())
print(text_2.capitalize())
```

```python
project_name = "machine learning project"
print(project_name.title())
course_name = "introduction to artificial intelligence"
print(course_name.title())
```
