```python
import numpy as np
import matplotlib.pyplot as plt
```

-  w controls steepness (how sharp the ramp is)
- b controls position (where the ramp goes up)

```python
#define smooth ramp(sigmoid function)
def sigmoid(x,w,b):
    return 1.0/(1.0+np.exp(-(w*x+b)))
```

```python
#Make 1 Tower by combining 2 sigmoid ramps

def make_tower(x,start_pos,end_pos,height=1.0,steepness=50.0):
    ramp1 = sigmoid(x,w = steepness,b= -steepness*start_pos)
    ramp2 = sigmoid(x,w = steepness,b= -steepness*end_pos)

    tower = height*(ramp1-ramp2)
    return tower
```

- Ramp 1 goes UP at start_pos
- Ramp 2 goes DOWN at end_pos

 Subtracting Ramp 2 from Ramp 1 leaves a single tower in between!

```python
# create X axis values(from -5 to 5)
x = np.linspace(-5,5,500)
```

```python
#build individual tower of different heights and location
tower1 = make_tower(x, start_pos=-4.0, end_pos=-2.0, height=0.8)
tower2 = make_tower(x, start_pos=-2.0, end_pos=0.0, height=-0.5)
tower3 = make_tower(x, start_pos=0.0, end_pos=2.0, height=1.0)
tower4 = make_tower(x, start_pos=2.0, end_pos=4.0, height=0.3)
```

```python
# Stack all towers together to approximate an arbitrary shape!
combined_towers = tower1 + tower2 + tower3 + tower4
```

```python
# Step 4: Plot the results
plt.figure(figsize=(10, 5))

# Plot the individual towers (Lego blocks)
plt.plot(x, tower1, '--', label='Tower 1', alpha=0.6)
plt.plot(x, tower2, '--', label='Tower 2', alpha=0.6)
plt.plot(x, tower3, '--', label='Tower 3', alpha=0.6)
plt.plot(x, tower4, '--', label='Tower 4', alpha=0.6)
```

```python
# Step 4: Plot the results
plt.figure(figsize=(10, 5))

# Plot the individual towers (Lego blocks)
plt.plot(x, tower1, '--', label='Tower 1', alpha=0.6)
plt.plot(x, tower2, '--', label='Tower 2', alpha=0.6)
plt.plot(x, tower3, '--', label='Tower 3', alpha=0.6)
plt.plot(x, tower4, '--', label='Tower 4', alpha=0.6)


# Plot the combined shape
plt.plot(x, combined_towers, color='black', linewidth=3, label='Combined Shape (Approximated Function)')

plt.title("Building Any Wave / Shape Using Towers (Sigmoid Neurons)")
plt.xlabel("Input (x)")
plt.ylabel("Output (y)")
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```
